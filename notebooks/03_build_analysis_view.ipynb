```python
import sqlite3
import pandas as pd

DB_PATH = "../data/cookie_cats.db"
conn = sqlite3.connect(DB_PATH)
```

```python
# Read the SQL file
with open("../sql/02_build_analysis_view.sql", "r") as f:
    sql_script = f.read()

# Execute it (SQLite needs executescript for multiple statements)
conn.executescript(sql_script)
conn.commit()
print("View created successfully")
```

```python
query = "SELECT COUNT(*) AS rows_in_view FROM analysis_view"
pd.read_sql(query, conn)
```

```python

query = "SELECT * FROM analysis_view LIMIT 5"
pd.read_sql(query, conn)
```

```python
query = """
SELECT
    engagement_segment,
    COUNT(*) AS users,
    ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) AS pct
FROM analysis_view
GROUP BY engagement_segment
ORDER BY users DESC
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT
    is_outlier,
    COUNT(*) AS users,
    MIN(sum_gamerounds) AS min_rounds,
    MAX(sum_gamerounds) AS max_rounds
FROM analysis_view
GROUP BY is_outlier
"""
pd.read_sql(query, conn)
```
